```python
import numpy as np
import pandas as pd

data = pd.read_excel('minor_survey.xlsx')
print(data.columns)                 # check hoe de kolommen precies heten

cv = lambda x: np.std(x, ddof=1) / np.mean(x) * 100   # ddof=1: het is een steekproef

print('CV length:', cv(data['length'].dropna()))
print('CV expected_grade:', cv(data['expected_grade'].dropna()))
```
