```python
"""
Insurance package problem
==========================

The company sells a COMBINED package: home insurance + car-liability insurance.

Given:
    P(home claim) = 0.06   (6% of homeowners claim)
    P(car claim)  = 0.05   (5% of car owners claim)

The company loses money on a customer if that customer claims on EITHER
policy (or both). So the company only makes a PROFIT if the customer
claims on NEITHER policy.

Assumption: whether someone claims on their home insurance and whether
they claim on their car insurance are INDEPENDENT events. (Nothing in the
problem suggests one type of claim makes the other more or less likely.)

Because the two events are independent, the probability that BOTH
"no home claim" and "no car claim" happen is just the product of the
two individual "no claim" probabilities:

    P(no home claim) = 1 - 0.06 = 0.94
    P(no car claim)  = 1 - 0.05 = 0.95

    P(profit) = P(no home claim AND no car claim)
              = P(no home claim) * P(no car claim)
              = 0.94 * 0.95
"""

# Step 1: probabilities of a claim on each policy
p_home_claim = 0.06
p_car_claim = 0.05

# Step 2: probabilities of NO claim on each policy (complement rule)
p_no_home_claim = 1 - p_home_claim
p_no_car_claim = 1 - p_car_claim

# Step 3: probability of profit = no claim on EITHER policy
# (independent events -> multiply the individual probabilities)
p_profit = p_no_home_claim * p_no_car_claim

# Step 4 (bonus): probability of a loss = the complement of profit
p_loss = 1 - p_profit

print(f"P(no home claim)      = {p_no_home_claim:.4f}")
print(f"P(no car claim)       = {p_no_car_claim:.4f}")
print(f"P(profit on customer) = {p_profit:.4f}  ->  {p_profit * 100:.2f}%")
print(f"P(loss on customer)   = {p_loss:.4f}  ->  {p_loss * 100:.2f}%")
```
