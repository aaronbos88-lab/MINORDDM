```python
from scipy.stats import beta, binom, betabinom

# --- Vul hier de gegevens uit je opgave in ---
alpha_prior = 1    # prior alpha
beta_prior  = 1    # prior beta
k = 0              # aantal geobserveerde harten
n = 0              # aantal geobserveerde trekkingen
# ---------------------------------------------

# Posterior parameters
a_post = alpha_prior + k
b_post = beta_prior + (n - k)
p_post = a_post / (a_post + b_post)          # posterior gemiddelde

print(f"Posterior: Beta({a_post}, {b_post})")
print(f"Posterior kans op harten: {p_post:.4f}")

# 1. Kans op precies 6 harten in 10 trekkingen (binomiaal met posterior p)
p_x6 = binom.pmf(6, 10, p_post)
print(f"P(X = 6) = {p_x6:.4f}")

# (Optioneel, volledig Bayesiaans: onzekerheid in p meenemen via beta-binomiaal)
p_x6_bb = betabinom.pmf(6, 10, a_post, b_post)
print(f"P(X = 6) beta-binomiaal = {p_x6_bb:.4f}")

# 2. Kans dat de kans op harten hoogstens 40% is (CDF van de posterior Beta)
p_le_40 = beta.cdf(0.40, a_post, b_post)
print(f"P(p <= 0.40) = {p_le_40:.4f}")
```
