```python
from sympy import symbols, diff, Eq, solve

s = symbols('s', positive=True)          # 1. variabele heet s i.p.v. x
C = 5*s + 0.2*s**3                       # 2. jouw functie i.p.v. x**2

dC_s = diff(C, s)                        # 3. afgeleide van C
print('De afgeleide is:', dC_s)          # -> 0.6*s**2 + 5

# De afgeleide bestaat uit een constant deel (5) en een kwadratisch deel (0.6*s**2).
# In plaats van dC_s = 0 op te lossen, zoek je waar deze twee termen gelijk zijn:
crossover = solve(Eq(0.6*s**2, 5), s)    # 4. i.p.v. solve(dx_f, x)
print('Energieverbruik versnelt sterk vanaf s =', crossover)
```

```python
import numpy as np
from scipy.integrate import quad
# Define the function daily_demand
def daily_demand(t):
    return 45 + 20 * np.sin((2 * np.pi / 7) * t)
# Compute the integral of the daily demand from day 0 to 30
integral, error = quad(daily_demand, 0, 7)
# Print the result
print(f'Total demand over one week: {integral} units')
```
