```python
#Exercise 1

import pandas as pd
WorldHappiness = pd.read_csv('World Happiness 2019.csv')
WorldHappiness.head()


```

```python
#Exercise 1a
import statsmodels.api as sm

y = WorldHappiness['Happiness_Score']
X = WorldHappiness[['GDP_per_capita','Social_support','Healthy_life_expectancy','Freedom_to_make_life_choices','Generosity','Perceptions_of_corruption']]
X = sm.add_constant(X) 

model1 = sm.OLS(y,X[['const','GDP_per_capita']]).fit()
print(model1.summary())
model2 = sm.OLS(y,X[['const','Social_support']]).fit()
print(model2.summary())
model3 = sm.OLS(y,X[['const','Healthy_life_expectancy']]).fit()
print(model3.summary())
model4 = sm.OLS(y,X[['const','Freedom_to_make_life_choices']]).fit()
print(model4.summary())
model5 = sm.OLS(y,X[['const','Generosity']]).fit()
print(model5.summary())
model6 = sm.OLS(y,X[['const','Perceptions_of_corruption']]).fit()
print(model6.summary())
```

```python
#Exercise 1b
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif

#Create a dataframe that contains the VIF values
VIF_data = pd.DataFrame() #create a pandas dataframe
VIF_data["variable"] = X.columns #create a variable that simply lists the names of the variables you picked
VIF_data["VIF"] = [vif(X,i) 
                   for i in range(len(X.columns))] #a loop that creates vif values for each variable and then puts it in the VIF column
print(VIF_data) #print the results
```

```python
#Exercise 1c
model7 = sm.OLS(y,X).fit()
print(model7.summary())
```

```python
#Exercise 3
import pandas as pd
GBvideos = pd.read_csv("GBvideos.csv")
GBvideos.head()

```

```python
#3b

#remove variables with no likes or dislikes
GBvideos = GBvideos[GBvideos.likes > 0] #interestingly, this one is not strictly necessary because there are no videos with > 0 dislikes and 0 likes
GBvideos = GBvideos[GBvideos.dislikes > 0]
        
#Remove rows which have blocked ratings, blocked comments, or have been removed
GBvideos = GBvideos[GBvideos.comments_disabled == False] #note that in the csv file it's spelled FALSE but in Python as False. This is because it is not a string variable but a boolean
GBvideos = GBvideos[GBvideos.ratings_disabled == False]
GBvideos = GBvideos[GBvideos.video_error_or_removed == False]
```

```python
#3c
#Create new variable
GBvideos = GBvideos.assign(like_percentage = GBvideos.likes/(GBvideos.likes+GBvideos.dislikes))

#check if there are no mistakes
GBvideos.head()
```

```python
#3d
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif 

X = GBvideos[['views','comment_count','like_percentage']]
X = sm.add_constant(X) #add a constant to the variables. VIF WON'T WORK CORRECTLY IF YOU DON'T DO THIS

#Create a dataframe that contains the VIF values
VIF_data = pd.DataFrame() #create a pandas dataframe
VIF_data["variable"] = X.columns #create a variable that simply lists the names of the variables you picked
VIF_data["VIF"] = [vif(X,i) 
                   for i in range(len(X.columns))] #a loop that creates vif values for each variable and then puts it in the VIF column
print(VIF_data) #print the results

```

```python
import statsmodels.api as sm

y = GBvideos['like_percentage']
X = GBvideos[['views','comment_count']]
X = sm.add_constant(X)
model = sm.OLS(y,X).fit()
print(model.summary())
```

```python
#Exercise 4
import pandas as pd
insurance = pd.read_csv('insurance.csv')
```

```python
#q4a: recode the variable
import numpy as np
insurance['east_west'] = np.where((insurance['region']=='southwest') | (insurance['region']=='northwest'), 1, 0)
#recode the categorical variables while we're at it
dummies = pd.get_dummies(insurance[['sex','smoker']],
                         dtype=int) #this turns them into 0s and 1s instead of true and false
insurance = pd.concat([insurance,dummies],axis=1)
insurance.head()
```

```python
#q4a: recode the variable
import numpy as np
insurance['east_west'] = np.where((insurance['region']=='southwest') | (insurance['region']=='northwest'), 1, 0)
#recode the categorical variables while we're at it
insurance = pd.get_dummies(insurance,dtype=int)
insurance.head()
```

```python
#q4b: check for multicollinearity
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif #import the VIF function


X = insurance[['age','bmi','children','east_west','sex_female','smoker_yes']] #select the relevant variables
X = sm.add_constant(X) #add a constant to the variables. VIF WON'T WORK CORRECTLY IF YOU DON'T DO THIS

#Create a dataframe that contains the VIF values
VIF_data = pd.DataFrame() #create a pandas dataframe
VIF_data["variable"] = X.columns #create a variable that simply lists the names of the variables you picked
VIF_data["VIF"] = [vif(X,i) 
                   for i in range(len(X.columns))] #a loop that creates vif values for each variable and then puts it in the VIF column
print(VIF_data) #print the results
```

```python
#q4c
import statsmodels.api as sm

y = insurance['charges']
X = insurance.drop('charges',axis=1)
X = sm.add_constant(X)

model1 = sm.OLS(y,X[['const','east_west']]).fit()
print(model1.summary())
```

```python
#q4d: 
model2 = sm.OLS(y,X[['const','east_west','age']]).fit()
print(model2.summary())
model3 = sm.OLS(y,X[['const','east_west','sex_female']]).fit()
print(model3.summary())
model4 = sm.OLS(y,X[['const','east_west','bmi']]).fit()
print(model4.summary())
model5 = sm.OLS(y,X[['const','east_west','children']]).fit()
print(model5.summary())
model6 = sm.OLS(y,X[['const','east_west','smoker_yes']]).fit()
print(model6.summary())
```
