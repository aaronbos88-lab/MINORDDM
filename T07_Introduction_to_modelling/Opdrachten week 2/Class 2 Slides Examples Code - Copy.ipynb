```python
#Example 1 with stork data
import pandas as pd
import statsmodels.api as sm

StorkData = pd.read_csv('StorkData.csv')

y = StorkData['BirthRate'] #extract the dependent variable from the data

#model1
X1 = StorkData['StorkPairs']
X1 = sm.add_constant(X1)
model1 = sm.OLS(y, X1).fit() #fit the model to the data
print(model1.summary()) #print the results

#model2
X2 = StorkData[['StorkPairs','LandArea']] #extract the independent variables
X2 = sm.add_constant(X2) #add an intercept 
model2 = sm.OLS(y, X2).fit() #fit the model to the data
print(model2.summary()) #print the results
```

```python
#Categorical variables
dataset_Facebook = pd.read_csv('dataset_Facebook.csv')

#create dummy variables
dataset_Facebook = pd.get_dummies(dataset_Facebook, dtype=int)

#check if it worked
dataset_Facebook.head()
```

```python
#Predicting likes based on paids
import statsmodels.api as sm

y = dataset_Facebook['like']
X = dataset_Facebook['Payment_Paid']
X = sm.add_constant(X)

model = sm.OLS(y,X).fit()
print(model.summary())
```

```python
#Regression with Type
y = dataset_Facebook['Lifetime_Engaged_Users']
X = dataset_Facebook[['Type_Status','Type_Video','Type_Link']]
X = sm.add_constant(X)

model2 = sm.OLS(y,X).fit()
print(model2.summary())
```

```python
#Example 3
import pandas as pd
UniversitySalaries = pd.read_excel('University Salaries.xlsx')
UniversitySalaries.head()

import statsmodels.api as sm
y = UniversitySalaries['Salary']
X = UniversitySalaries[['Age','Experience']]
X = sm.add_constant(X)

model1 = sm.OLS(y,X[['const','Age']]).fit()
print(model1.summary())
model2 = sm.OLS(y,X[['const','Experience']]).fit()
print(model2.summary())
model3 = sm.OLS(y,X).fit()
print(model3.summary())
```

```python
#VIF
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif #import the VIF function
from statsmodels.tools.tools import add_constant #import the add_constant function

X = UniversitySalaries[['Age','Experience','Student_Approval']] #select the relevant variables and delete missing values
X = add_constant(X) #add a constant to the variables. VIF WON'T WORK CORRECTLY IF YOU DON'T DO THIS

#Create a dataframe that contains the VIF values
VIF_data = pd.DataFrame() #create a pandas dataframe
VIF_data["variable"] = X.columns #create a variable that simply lists the names of the variables you picked
VIF_data["VIF"] = [vif(X,i) 
                   for i in range(len(X.columns))] #a loop that creates vif values for each variable and then puts it in the VIF column
print(VIF_data) #print the results


```

```python
#VIF with the video games dataset
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif #import the VIF function
from statsmodels.tools.tools import add_constant #import the add_constant function

import pandas as pd
video_games = pd.read_excel('video_games.xlsx')

X = video_games[['Review_Score','Used_Price','Release_Year','Length_All_PlayStyles_Polled']].dropna() #select the relevant variables and delete missing values
X = add_constant(X) #add a constant to the variables. VIF WON'T WORK CORRECTLY IF YOU DON'T DO THIS

#Create a dataframe that contains the VIF values
VIF_data = pd.DataFrame() #create a pandas dataframe
VIF_data["variable"] = X.columns #create a variable that simply lists the names of the variables you picked
VIF_data["VIF"] = [vif(X,i) 
                   for i in range(len(X.columns))] #a loop that creates vif values for each variable and then puts it in the VIF column
print(VIF_data) #print the results

```
