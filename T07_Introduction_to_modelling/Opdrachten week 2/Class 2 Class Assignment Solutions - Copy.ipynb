```python
#In class exercise 1(a)
import pandas as pd
video_games = pd.read_excel('video_games.xlsx')
video_games.head()
```

```python
#In class exercise 1(b)
import statsmodels.api as sm

y = video_games['Sales']

X = video_games.drop('Sales',axis=1) #this code drops the Sales variable so that you are left with all the other variables. The axis=1 part causes you to drop a column and not a row.
X = sm.add_constant(X)

model1 = sm.OLS(y,X[['const','Review_Score']]).fit() #note that the 'const' variable is the intercept
model2 = sm.OLS(y,X[['const','Used_Price']]).fit() 
model3 = sm.OLS(y,X[['const','Release_Year']]).fit() 
model4 = sm.OLS(y,X[['const','Length_All_PlayStyles_Polled']]).fit() 

print(model1.summary())
print(model2.summary())
print(model3.summary())
print(model4.summary())
```

```python
#In class exercise 1(c)
model5 = sm.OLS(y,X).fit()
print(model5.summary())
```

```python
#Class Exercise 2(a)
import pandas as pd

insurance = pd.read_csv('insurance.csv')
insurance = pd.get_dummies(insurance, dtype=int)
insurance.head()
```

```python
#Class Exercise 2(b)
import statsmodels.api as sm

y = insurance['charges']
X = insurance['sex_male']
X = sm.add_constant(X)

model = sm.OLS(y,X).fit()
print(model.summary())
```

```python
#Class Exercise 2(c)
import statsmodels.api as sm

y = insurance['charges']
X = insurance[['sex_male','smoker_yes']]
X = sm.add_constant(X)

model = sm.OLS(y,X).fit()
print(model.summary())
```
