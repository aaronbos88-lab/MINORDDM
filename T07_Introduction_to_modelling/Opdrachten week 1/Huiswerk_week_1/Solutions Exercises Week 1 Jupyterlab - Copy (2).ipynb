```python
#import the data
import pandas as pd
WorldHappiness = pd.read_csv('World Happiness 2019.csv')
WorldHappiness.head()
```

```python
#Exercise 1a
import matplotlib.pyplot as plt

Happiness_Score = WorldHappiness['Happiness_Score']
GDP_per_capita = WorldHappiness['GDP_per_capita']

plt.scatter(Happiness_Score,GDP_per_capita)
plt.title('Happiness and GDP')
plt.xlabel('GDP per capita')
plt.ylabel('Happiness Score')
```

```python
#Exercise 1b
import scipy.stats

scipy.stats.pearsonr(Happiness_Score,GDP_per_capita)
```

```python
#Exercise 1c
import seaborn as sn
data = WorldHappiness[['Happiness_Score','GDP_per_capita','Social_support','Healthy_life_expectancy',
                       'Freedom_to_make_life_choices','Generosity','Perceptions_of_corruption']] #select only the relevant data
sn.pairplot(data) #create pairplot

matrix = data.corr() #create correlation matrix
sn.heatmap(matrix, annot=True) #create heatmap

scipy.stats.pearsonr(data['Happiness_Score'],data['Generosity']) #there is no significant correlation between happiness and generosity
```

```python
#Exercise 2

#import file
video_games = pd.read_excel('video_games.xlsx')
video_games.head()

#import relevant package for the regression
import statsmodels.api as sm

```

```python
#model1
#note: the codes for the four models can be made much shorter. Can you figure out how?
y = video_games['Sales']
X = video_games['Review_Score']
X = sm.add_constant(X)

model1 = sm.OLS(y,X)
results = model1.fit()
print(results.summary())
```

```python
#model2
y = video_games['Sales']
X = video_games['Used_Price']
X = sm.add_constant(X)

model2 = sm.OLS(y,X)
results = model2.fit()
print(results.summary())
```

```python
#model3
y = video_games['Sales']
X = video_games['Release_Year']
X = sm.add_constant(X)

model3 = sm.OLS(y,X)
results = model3.fit()
print(results.summary())
```

```python
#model4
y = video_games['Sales']
X = video_games['Length_All_PlayStyles_Polled']
X = sm.add_constant(X)

model4 = sm.OLS(y,X)
results = model4.fit()
print(results.summary())
```
